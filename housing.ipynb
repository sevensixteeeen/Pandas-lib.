```python
import pandas as pd
```

```python
df = pd.read_csv("housing.csv", index_col=0)
df = df.drop_duplicates()
print(df)
```

```python
print(df.info())
```

```python
print("total flats", len(df))
print("Average price:", df["price"].mean())
print("Maximum price:", df["price"].max())
print("Maximum bedrooms:", df["bedrooms"].max())
print("Minimum bedrooms:", df["bedrooms"].min())
print("Minimum price:", df["price"].min())
```

Data cleaning 

```python
df["date"] = pd.to_datetime(df["date"])
df[df.index.duplicated(keep=False)]
df[df.bedrooms > 10]
```

```python
df.loc[df["bedrooms"] == 33, "bedrooms"] = 3
```

```python
df = df[(df["bedrooms"] > 0) & (df["bathrooms"] > 0)].copy()
```

```python
print("total flats", len(df))
print("Maximum bedrooms:", df["bedrooms"].max())
print("Minimum bedrooms:", df["bedrooms"].min())
print("Minimum bathrooms:", df["bathrooms"].min())
```
