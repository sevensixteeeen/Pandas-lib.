Filtering 

```python
tall_pokemon = df[df['Height'] > 2]  # to see all the rows where height is greater than 2 use df[df['Height'] > 2]
print(tall_pokemon)
```

```python
water_pokemon=df[(df['Type1']=='Water')  | 
                 (df['Type2']=='Water')]  # to see all the rows where type1 or type2 is water use df[df['type1']=='water'|df['type2']=='water']

print(water_pokemon)
```

```python
ff_pokremon = df[(df['Type1']=='Fire') & (df['Type2']=='Flying')]  # to see all the rows where type1 is fire and type2 is flying use df[df['type1']=='fire'&df['type2']=='flying'] 

print(ff_pokremon)
```

```python
wg_pokemon = df[((df['Type1']=='Water') & (df['Type2']=='Grass')) |
                ((df['Type1']=='Grass') & (df['Type2']=='Water'))]  # check both orders: water/grass or grass/water
print(wg_pokemon)
print("number of water and grass pokemon:", len(wg_pokemon))  # 0, because no gen 1 pokemon is water and grass both.
```
