```python
df = pd.read_csv('pdata.csv', index_col='Name')  # to read a csv file use pd.read_csv('file_name.csv'). Use index_col to set a column as index.
print(df)  # too see all the rows and columns use .to_string() after df.
# print(df.shape)  # to see the number of rows and columns use df.shape.
```

```python
# print(df["Name"])  # to see a particular column use df["column_name"]

print(df[["Name", 'Type1','Weight']])  # to see multiple columns use df[["column_name1", "column_name2"]]
```

```python
# to see the first 5 rows use df.head(). You can also pass a number to see that many rows.

print(df.head())  
```

```python
print(df.loc['Pikachu'])  # to see a particular row use df.loc[row_index]
```

```python
pokemon = input("Enter the name of the pokemon: ")
try:
    print(df.loc[pokemon])  # to see a particular row use df.loc[row_index]
except KeyError:
    print("Pokemon not found in the dataset.")
```

Filtering 

```python
tall_pokemon = df[df['Height'] > 2]  # to see all the rows where height is greater than 2 use df[df['Height'] > 2]
print(tall_pokemon)
```

```python
water_pokemon=df[(df['Type1']=='Water')  | 
                 (df['Type2']=='Water')]  # to see all the rows where type1 or type2 is water use df[df['type1']=='water'|df['type2']=='water']

print(water_pokemon)
```

```python
ff_pokremon = df[(df['Type1']=='Fire') & (df['Type2']=='Flying')]  # to see all the rows where type1 is fire and type2 is flying use df[df['type1']=='fire'&df['type2']=='flying'] 

print(ff_pokremon)
```
