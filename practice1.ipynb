```python
# calories counter analogy with a dictionary

calories_data = {'day 1': 420, 'day 2': 380, 'day 3': 360, 'day 4': 400, 'day 5': 450}

series_calories = pd.Series(calories_data)
print(series_calories)
```

```python
# calories counter analogy with a dictionary

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

```python
# accessing loc propeties in dictionary
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar.loc["day1"])
```

```python
# filtering a pandas Series based on multiple conditions
d=[100,200,50,300,100,250,275,500]

series_d=pd.Series(d,index=['a','b','c','d','e','f','g','h'])
print(series_d)
print("Filtering series_d for values greater than 200 and less than 300:")
print(series_d[(series_d>200) & (series_d<300)])
```

```python
# updating series
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

myvar.loc['day3'] += 500

print(myvar.loc["day3"])
```

```python
#creating series
data=[100,200,300,400,500]
print("data before creating series:",data)
series=pd.Series(data)
print("series created:",series)
```

Data frame:  A tabular data structure with rows and columns.



```python
data ={
    "name":["Ravi","Meera","Zoya"],
    "age":[25,30,28],
    "city":["Delhi","Pune","Delhi"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# indexing
data ={
    "name":["Ravi","Meera","Zoya"],
    "age":[25,30,28],
    "city":["Delhi","Pune","Delhi"]
}
df = pd.DataFrame(data,index=["employee 1", "employee 2", "employee 3"] )
print(df)
```

```python
data ={
    "name":["Ravi","Meera","Zoya"],
    "age":[25,30,28],
    "city":["Delhi","Pune","Delhi"]
}
df = pd.DataFrame(data,index=["employee 1", "employee 2", "employee 3"] )
print(df.loc["employee 1"])
# asscessing it by iloc properties
print(df.iloc[1])
```

```python
data ={
    "name":["Ravi","Meera","Zoya"],
    "age":[25,30,28],
    "city":["Delhi","Pune","Delhi"]
}
df = pd.DataFrame(data,index=["employee 1", "employee 2", "employee 3"] )
# add a new column
df["salary"] = [1000,2000,3000]

# add a new row
new_rows= pd.DataFrame([{"name": "ajay", "age": 32, "city":"pune", "salary": 30000 },
                        {"name": "recoup", "age": 46, "city":"channai", "salary": 40000 }],
                      index=["employee 4","employee 5"])
df=pd.concat([df,new_rows])
print(df)
```
